```python
#데이터 이슈 3 해결
data_set=[[270,0,0],[250,0,0],[220,1,1],[240,1,1]]
df=pd.DataFrame(data=data_set,columns=['무게','표면상태','결과'])
df
```

```python
X=df.drop('결과',axis=1).to_numpy()
y=df.drop(['무게',	'표면상태'],axis=1).to_numpy()
```

```python
#3.학습
from sklearn.tree import DecisionTreeClassifier
m=DecisionTreeClassifier()
m.fit(X,y)
```

```python
#4. 예측 및 검증
m.score(X,y)
```

```python
#5. 모델 사용 예측
def 과일_판별기(x1,x2):
    모델_입력데이터=[]
    입력데이터=[x1,x2]
    모델_입력데이터.append(입력데이터)
    결과_집합=m.predict(모델_입력데이터)
    결과=결과_집합[0]
    정답=['오렌지','사과']
    return 정답[결과]
과일_판별기(220,1)
```
