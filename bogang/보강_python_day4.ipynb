# 다형성
1. 오버로딩: 같은 공간에 다른 대상을 만드는 작업
2. 오버라이딩: 다른공간에 같은 대상을 만드는 작업

```python
class B:
    def __init__(self):
        print("B의 생성자 ")
    def f(self):
        print("인스턴스 함수")
```

```python
a=B()
a.f()
```

```python
b=B()
```

```python
class A(B):
    def __init__(self):#오바라이딩
        super().__init__()
        pass
b=A()
```

```python
b.f()
```

```python
class C:
    def __init__(self,data):
        self.data=data
    def __add__(self,x):
        return self.data-x.data
```

```python
a=C(10)
b=C(100)
a.__add__(b)
a+b
```

```python
class A:
    __x=10
    x=10
    @classmethod
    def f(cls):
        print(cls.__x)
#A.__x
A.f()
```

```python
class B(A):
    @classmethod
    def f(cls):
        print(cls.x)
B.f()   
```

```python
class A:
    def f(self):
        print("A의 메서드")
class B:
    def f(self):
        print("B의 메서드")
class C(A):
    def f(self):
        print("C의 메서드")
```

```python
class D(C):
    pass
a=D()
a.f()
```

```python
class E(B,A):
    pass
b=E()
b.f()
```
