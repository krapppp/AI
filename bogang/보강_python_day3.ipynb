```python
#조건
for 1 in a_l:
    pass
else:
    print("프로그램 동작후 자원정리")
while True:
    pass
else:
    print("프로그램 동작후 자원정리")
```

# 블록(실행/제어)
- 들여쓰기로 판단
- 일반적인 블록은 전부다 실행블록이다.
- 제어블록은 반복문블록 이다.
- 실행블록은 무조건 위에서 아래로 진행된다.
- 제어블록은 위로도 이동이 가능하다.

```python
for 변수 in range(5):
    print(변수)
```

```python
range(5)#집합구조
```

```python
range(시작값=0,종료조건,증감식=1)
```

```python
list(range(10))
```

```python
list(range(10,0,-1))
```

```python
list(range(1,10))
```

```python
[0 for _ in range(10)]
```

```python
[0]*10
```

```python
a=[[0,0,0,0,0]]*5
a[0][0]=10
a
```

```python
b=[[0 for _ in range(5)] for _ in range(5)]
b[0][0]=10
b
```
