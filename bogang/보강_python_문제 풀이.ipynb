```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("함수 실행 전 동작")
        result = func(*args, **kwargs)
        print("함수 실행 후 동작")
        return result
    return wrapper

@my_decorator
def say_hello(name):
    print(f"안녕하세요, {name}님!")

say_hello("철수")
```

```python
from random import randint
def game():
    win,n,c_win,ck=[0 for _ in range(4)]
    d={'가위':0,'바위':1,'보':2}
    str_data=['가위','바위','보']
    while True:
        P_in=input("입력:")
        if P_in not in str_data:
            print("잘못입력")
            continue
        P_in=d[P_in]
        C_in=randint(0,2)
        print(f'컴퓨터: {str_data[C_in]} 플래이어: {str_data[P_in]}',end=' ')
        if ((P_in == 0 and C_in == 1) or
                (P_in == 1 and C_in == 2) or
                (P_in == 2 and C_in == 0)):
            n+=1
            print('패배')
            print('패배로 인한 ')
            return f"결과\n게임수: {n}회 승리: {win}회 연승: {c_win}회 "
        elif P_in==C_in:
            n+=1
            ck = False
            print("비김")
        else:
            n+=1
            if ck:
                c_win+=1
            ck = True
            win+=1
            print("승리")
print('가위,바위,보 게임을 시작합니다.')
print(game())
```

```python
def f(n):
    out_d=''
    for i in n:
        if i.islower():
            out_d += i.upper()
        elif i.isupper():
            out_d += i.lower()
        elif ' '==i:
            out_d += ' '
        else:
            print(f'출력: {out_d}\n')
            return True
    else:
        print(f'출력: {out_d}\n')
run=False
while not run:
    run=f(input("입력: "))
else:
    print('프로그램이 종료됩니다.')
```

```python
def base2(n):
	if n < 2:
		print(numberChar[n], end = ' ')
	else:
		base2(n // 2)
		print(numberChar[n % 2], end = ' ')
def base8(n):
	if n < 8:
		print(numberChar[n], end = ' ')
	else:
		base8(n // 8)
		print(numberChar[n % 8], end = ' ')
def base16(n):
	if n < 16:
		print(numberChar[n], end = ' ')
	else:
		base16(n // 16)
		print(numberChar[n % 16], end = ' ')


numberChar = '0123456789ABCDEF'

number = int(input('10진수 입력 -->'))
print('\n 2진수 : ', end = ' ')
base2(number)
print('\n 8진수 : ', end = ' ')
base8(number)
print('\n16진수 : ', end = ' ')
base16(number)
```

```python
from random import randint
w,d,l=[0 for _ in range(3)]
n=int(input('게임횟수를 입력하시오:'))
print('_____________________________________________')
print(f'게임을 {n}회 시작합니다.')
ck_d={'ck_p':[],'ck_c':[],'ck':[]}
for i in range(1,n+1):
    C_in=randint(1,6)
    ck_d['ck_c'].append(C_in)
    print(f'_____________________________________________\n{i}회차\n컴퓨터 : {C_in}')
    input('주사위를 굴리시오')
    P_in = randint(1, 6)
    ck_d['ck_p'].append(P_in)
    print(f'플래이어 : {P_in}\n_____________________________________________')
    if P_in>C_in:
        ck_d['ck'].append('승리')
        w+=1
    elif P_in<C_in:
        ck_d['ck'].append('패배')
        l += 1
    elif P_in==C_in:
        ck_d['ck'].append('비김')
        d += 1

print(f'결과\n{n}번 경기내용')
for i in range(n):
    print(f'{i+1}회차 - 플래이어: {ck_d["ck_p"][i]}, 컴퓨터: {ck_d["ck_c"][i]} {ck_d["ck"][i]}')
else:
    print(f'''_____________________________________________
게임수: {n}회 승리: {w}회 비김: {d}회 패배{l}회
_____________________________________________''')
```

```python
import pickle
class 금액부족(Exception):
    def __str__(self):
        return "통장에 잔액이 부족합니다."

class Account:
    account = {}

    def __init__(self, ano, owner, balance):
        self.__ano = ano
        self.__owner = owner
        self.__balance = balance
        self.__record = f"{'초기금액:' + str(balance):<10s}|{'합계금액:' + str(balance):<10s}|\n"

    def getAno(self):
        return self.__ano

    def setAno(self, ano):
        self.__ano = ano

    def getOwner(self):
        return self.__owner

    def setOwner(self, owner):
        self.__owner = owner

    def getBalance(self):
        return self.__balance

    def setBalance(self, balance):
        self.__balance = balance

    def getrecord(self):
        return self.__record

    def recording(self, changedBalance):
        if (changedBalance >= 0):
            self.__record += (f"{'입금내역:' + str(changedBalance):<10s}|{'합계금액:' + str(self.__balance):<10s}|")
        else:
            self.__record += (f"{'출금내역:' + str(changedBalance * -1):<10s}|{'합계금액:' + str(self.__balance):<10s}|")
        self.__record += "\n"

    @classmethod
    def createAccount(cls):
        print("--------------")
        print("계좌생성")
        print("--------------")
        ano = input("계좌번호: ")
        owner = input("계좌주: ")
        balance = int(input("초기입금액: "))
        newAccount = cls(ano, owner, balance)

        if ano in cls.account:
            print("이미 있는 계좌 입니다.")
        else:
            cls.account[ano] = newAccount
            print("계좌를 등록했습니다.")

    @classmethod
    def accountList(cls):
        print("--------------")
        print("계좌목록")
        print("--------------")
        for key, account in cls.account.items():
            print(account.getAno(), end="\t|")
            print(account.getOwner(), end="\t|")
            print(account.getBalance())

    @classmethod
    def deposit(cls):
        print("--------------")
        print("예금")
        print("--------------")
        ano = input("계좌번호: ")
        money = int(input("예금액: "))
        if ano not in cls.account:
            print("결과: 계좌가 없습니다.")
            return
        account = cls.account.get(ano)
        account.setBalance(account.getBalance() + money)
        account.recording(money)
        print("결과: 예금이 성공되었습니다.")

    @classmethod
    def withdraw(cls):
        print("--------------")
        print("출금")
        print("--------------")
        ano = input("계좌번호: ")
        money = int(input("출금액: "))
        if ano not in cls.account:
            print("결과: 계좌가 없습니다.")
            return
        account = cls.account.get(ano)
        if account.getBalance() < money:
            raise 금액부족()
        account.setBalance(account.getBalance() - money)
        money *= -1
        account.recording(money)
        print("결과: 출금이 성공되었습니다.")

    @classmethod
    def remit(cls):
        print("--------------")
        print("송금")
        print("--------------")
        ano1 = input("출금 계좌번호: ")
        ano2 = input("입금 계좌번호: ")
        remit_money = int(input("출금액: "))
        if (remit_money > 0):
            if ano1 not in cls.account:
                print("결과: 계좌가 없습니다.")
                return
            if ano2 not in cls.account:
                print("결과: 계좌가 없습니다.")
                return
            account = cls.account.get(ano1)
            if account.getBalance() < remit_money:
                raise 금액부족()
            account.setBalance(account.getBalance() - remit_money)
            remit_money *= -1
            account.recording(remit_money)
            remit_money *= -1
            account = cls.account.get(ano2)
            account.setBalance(account.getBalance() + remit_money)
            account.recording(remit_money)
            print("결과: 송금이 성공되었습니다.")
        else:
            print("송금액은 양수여야함.")

    @classmethod
    def showRecord(cls):
        print("--------------")
        print("내역")
        print("--------------")
        find_ano = input("조회 계좌번호: ")
        if find_ano not in cls.account:
            print("결과: 조회계좌가 없습니다.")
            return
        account = cls.account.get(find_ano)
        print(f"{'거래내역':^10s}|{'합계금액':^10s}|")
        print(account.getrecord())

    @classmethod
    def deleteAno(cls):
        print("--------------")
        print("삭제")
        print("--------------")
        find_ano = input("삭제 계좌번호: ")
        if find_ano not in cls.account:
            print("결과: 조회계좌가 없습니다.")
            return
        ck=input("등록된 계좌가 있습니다.\n삭제 하시겠습니까?\n<삭제:1,취소:자율입력>")[0]
        if ck=='1':
            del cls.account[find_ano]
        else:
            print('삭제를 취소 하였습니다.')

    @classmethod
    def endProgram(cls):
        print('프로그램을 종료합니다.')
        return True
ui="""-----------------------------------------------------------------
1.계좌생성 | 2.계좌목록 | 3.예금 | 4.출금 | 5.종료 | 6.송금 | 7.내역 | 8.삭제
-----------------------------------------------------------------------
선택>"""
model=[Account.createAccount,Account.accountList,Account.deposit,Account.withdraw
    ,Account.endProgram,Account.remit,Account.showRecord,Account.deleteAno]
run = False
try:
    open("save_data.p", "rb")
except:
    print("기존에 저장된 data가 없습니다.")
else:
    print('저장된 data를 로드했습니다.')
    Account.account = pickle.load(open("save_data.p", "rb"))
while not run:
    selectNo = input(ui)
    if not('1'<=selectNo<='8'):
        print('잘못입력하셨습니다.\n기능은 1~8까지만 있습니다.')
        continue
    selectNo=int(selectNo[0])
    run=model[selectNo-1]()
pickle.dump(Account.account, open("save_data.p", "wb"))
```
