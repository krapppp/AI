# 연산자
- 연산자의 우선순위와 방향
- 연산은 일반적으로 같은 자료형 끼리 연산한다.
- 연산은 한번에 하나의 결과만 만든다.
1. 산술연산자 +,-,*,/(몫연산),%(나머지)
   - 입력: 숫자 -> 출력:숫자
   - 왼 -> 오 
2. 관계(조건)연산자 (동등/대소) <정답 말하기>
   - 입력:숫자 -> 출력:논리, 입력:논리 -> 출력:논리
   - 왼 -> 오
3. 논리 연산자
   - 입력:논리->출력:논리
   - 왼 -> 오
4. 대입 연산자
   - (공간의 자리) = (값의 자리)
   - 왼 <- 오
   - 가장 느린 연산자 

```python
a=10
b=20
a == b #(동등 관계연산) False
a != b #(동등 관계연산) True
```

```python
a = 10+20 >100 and 20*10 <200 
```

```python
#(공간) = (값)
a=10
a(a이름의 공간) = a(a가 가지고 있는 값) 
```
