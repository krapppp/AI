# Q9 전체-배치 경사하강법(full-batch gradient descent) 방식과 미니-배치 경사하강법(minibatch gradient descent) 방식의 차이점에 대해서 바르게 설명한 것을 고르시오.
## ① 전체-배치 경사하강법은 모든 데이터를 한 번에 넣고 미니-배치 경사하강법은 데이터를 n개로 나누어 학습한다.
## ② 미니-배치 경사하강법에서 데이터를 나눌 때 일반적으로 10의 배수로 데이터를 나눈다.
## ③ 전체-배치 경사하강법을 할 때 데이터를 셔플(shuffle) 하게 되면, 데이터 입력의 일관성이 틀어져 학습 결과가 나빠진다.
## ④ 미니-배치 경사하강법에서는 1개의 데이터만 넣을 수 없다.

# Q10 회귀식의 성능을 측정하기 위해 사용되는 MAE, RMSE, 결정계수(R-squared)에 대한 설명으로 적절한 것을 고르시오. 
## ① RMSE는 예측값과 실제값의 차이에 대한 제곱을 모두 합한 후 제곱근 연산을 한 값에 음수곱을 취한 값이다.
## ② MAE는 예측값과 실제값 간의 차이에 대한 합을 데이터 개수n 으로 나눈 값으로 계산된다.
## ③ 결정계수(R-squared)는 두 값의 상관도를 지표화하여 -10에서 10까지의 값을 출력한다.
## ④ MAE는 0에 수렴할수록 좋고,  결정계수는 1에 수렴할수록 좋은 결과 이다. 이상치가 많을 경우에는 RMSE가 MAE보다 왜곡 정도를 좀 더 효과적으로 표현한다.

# Q11 경사하강법에서 학습률(learning rate)과 반복(iteration) 횟수가 중요한 하이퍼 매개변수인 이유로 맞는 설명을 고르시오. 
## ① 학습률이 크면 클수록 학습의 속도가 빨라져 최종 성능이 높아진다.
## ② 반복이 많을수록 최종 성능이 높아진다.
## ③ 학습률이 너무 작을 경우 학습 시간이 늦어지고, 반복이 충분하지 않을 경우 손실(loss)이 수렴되지 않는다.
## ④ 학습률이 작으면 세부적인 학습이 가능하며, 학습의 정확도를 위해 작게 하는 것이 좋다.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from keras.models import Sequential,Model
from keras.layers import Input,Dense,Conv2D,MaxPooling2D,Dropout,SimpleRNN,LSTM,GRU,BatchNormalization,Flatten
from keras.optimizers import Adam,RMSprop,SGD,Adagrad
from keras.losses import binary_crossentropy,categorical_crossentropy,sparse_categorical_crossentropy,mean_squared_error
from keras.callbacks import EarlyStopping,ModelCheckpoint
from keras.models import load_model
from sklearn.metrics import r2_score,classification_report,f1_score,mean_squared_error
```

```python
#1회 실행후 진행
import numpy as np
X=np.load('CNN_RNN_X.npy')
y=np.load('CNN_RNN_y.npy')
```

```python
X,X.shape
```

```python
y,y.shape
```

```python
#코드 작성 위치
from keras.layers import Flatten

tr_x,val_x,tr_y,val_y=train_test_split(X,y,train_size=0.8,stratify=y)

m=Sequential()
m.add(Input(shape=tr_x.shape[1:]))
m.add(Flatten())
m.add(Dense(100,activation='sigmoid'))#은닉
m.add(Dense(100,activation='sigmoid'))#은닉
m.add(Dense(10,activation='sigmoid'))#출력

m.compile(optimizer='sgd',loss='sparse_categorical_crossentropy',metrics=['acc'])
m.fit(tr_x,tr_y,validation_data=(val_x,val_y),epochs=100)
m.evaluate(tr_x,tr_y)
```

# P-Q2 데이터셋을 이용하여 CNN 기반의 분류 모델을 구현하시오.
## 요구사항

### 4. model구조 출력

```python
#1회 실행후 진행
import numpy as np
X=np.load('CNN_RNN_X.npy')
y=np.load('CNN_RNN_y.npy')
```

```python
#코드 작성 위치
from keras.layers import Flatten
from keras.datasets import fashion_mnist
from keras.layers import Rescaling
from keras.metrics import Accuracy,F1Score
from sklearn.metrics import classification_report
f1sc=F1Score()

tr_x,val_x,tr_y,val_y=train_test_split(X,y,test_size=0.3,stratify=y)

m=Sequential()
m.add(Input(shape=tr_x.shape[1:]))#(32,32,3)
m.add(Conv2D(10,3,1,padding='same',activation='relu'))#(32,32,10)
m.add(MaxPooling2D(2))#(16,16,10)
m.add(Conv2D(10,3,1,padding='same',activation='relu'))#(16,16,10)
m.add(MaxPooling2D(2))#(8,8,10)
m.add(Conv2D(10,3,1,padding='same',activation='relu'))#(8,8,10)
m.add(MaxPooling2D(2))#(4,4,10)
m.add(Conv2D(10,3,1,padding='same',activation='relu'))#(4,4,10)
m.add(MaxPooling2D(2))#(2,2,10)
m.add(Conv2D(10,3,1,padding='same',activation='relu'))#(2,2,10)
m.add(MaxPooling2D(2))#(1,1,10)
m.add(Flatten())
m.add(Dense(10,activation='softmax'))
m.summary()

m.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['acc',f1sc])
hy=m.fit(tr_x,tr_y,epochs=10)
```

```python
#코드 작성 위치

import os, cv2, numpy as np
import random
from keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
from keras.applications.vgg16 import VGG16, preprocess_input as vgg_preprocess, decode_predictions as vgg_decode
from keras.applications import ResNet50 as ResNet50_base
from keras.models import Model
from keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator

resnet_model = ResNet50(weights='imagenet')
vgg_model = VGG16(weights='imagenet')

dc=['/cat','/dog']

for x in dc:
    img_dir = 'C:/Users/ksa/Desktop/vs/pn/deep/test/cats_vs_dogs_small/train' + x
    all_files = [f for f in os.listdir(img_dir) if f.lower().endswith(('.jpg','.png','.jpeg'))]
    sample_files = random.sample(all_files, 10)

    for fname in sample_files:
        img_path = os.path.join(img_dir, fname)
        img = cv2.imread(img_path)
        if img is None:
            continue

        data = cv2.resize(img, (224,224)).reshape(1,224,224,3)

        resnet_input = preprocess_input(np.copy(data))
        resnet_preds = resnet_model.predict(resnet_input)
        resnet_output = decode_predictions(resnet_preds)

        vgg_input = vgg_preprocess(np.copy(data))
        vgg_preds = vgg_model.predict(vgg_input)
        vgg_output = vgg_decode(vgg_preds)

        res_label = f"ResNet50: {resnet_output[0][0][1]} {resnet_output[0][0][2]:.2f}"
        vgg_label = f"VGG16: {vgg_output[0][0][1]} {vgg_output[0][0][2]:.2f}"

        cv2.putText(img, res_label, (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 1)
        cv2.putText(img, vgg_label, (10, 40), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,255), 1)
        cv2.imshow('img', img)
        key = cv2.waitKey(0)
        if key == 27:  # ESC 누르면 종료
            break

    cv2.destroyAllWindows()
```
