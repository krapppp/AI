# 산술 연산
### 가중치 합
- 기존 덧셈 결과가 255보다 크면 픽셀값 255로 설정하는 saturate 기능으로 인해 영상의 많은 부분이 흰색으로 표현
- 조정 위해 가중치 합 적용
- $\text{dst}(x,y) = \text{saturate}(\alpha\cdot\text{src1}(x,y) + \beta\cdot\text{src2}(x,y))$<br>
    -> $\alpha + \beta = 1$ 되도록 설정
- cv2.addWeighted(src1, a, src2, b, gamma, dst = None, dtype = Mpme)
    - src1 : 첫번째 영상 or 스칼라
    - a : 첫번째 영상 가중치
    - src2 : 두번째 영상 or 스칼라
    - b : 두번째 영상 가중치
    - gamma : 결과 영상에 추가적으로 더할 값
    - dst : 연산 결과 영상
    - dtype : 출력 영상 타입 (e,g) -> cv2.CV_8U, cv2.CV_32F 등

```python
from matplotlib import pyplot as plt

src1=cv2.imread('./img/cat.jpg',cv2.IMREAD_GRAYSCALE)
src2=cv2.imread('./img/dog.jpg',cv2.IMREAD_GRAYSCALE)

dst1=cv2.add(src1,src2)
dst2=cv2.addWeighted(src1,0.8,src2,0.2,0)
dst3=cv2.addWeighted(src1,0.5,src2,0.5,0)
dst4=cv2.addWeighted(src1,0.2,src2,0.8,0)

plt.subplot(231), plt.axis('off'), plt.imshow(src1,'gray'), plt.title('cat')
plt.subplot(232), plt.axis('off'), plt.imshow(src2,'gray'), plt.title('dog')
plt.subplot(233), plt.axis('off'), plt.imshow(dst1,'gray'), plt.title('add')
plt.subplot(234), plt.axis('off'), plt.imshow(dst2,'gray'), plt.title('0.8_0.2')
plt.subplot(235), plt.axis('off'), plt.imshow(dst3,'gray'), plt.title('0.5_0.5')
plt.subplot(236), plt.axis('off'), plt.imshow(dst4,'gray'), plt.title('0.2_0.8')
```

### 뺄셈 연산
- 두 영상의 같은 위치 존재 픽셀 값에 대해 뺄셈 연산 수행
- 결과 < 0 : 픽셀값 0 설정
- $\text{dst}(x,y) = \text{saturate}(\text{src1}(x,y) - \text{src2}(x,y))$
- cv2.subtract(src1, src2, dst = None, mask = None, dtype = Mpme)
    - src1 : 첫번째 영상 or 스칼라
    - src2 : 두번째 영상 or 스칼라
    - dst :  뺄셈 연산 결과
    - mask : 마스크 영상
    - dtype : 출력 영상 타입 (e,g) -> cv2.CV_8U, cv2.CV_32F 등

```python
from matplotlib import pyplot as plt

src1=cv2.imread('./img/cat.jpg',cv2.IMREAD_GRAYSCALE)
src2=cv2.imread('./img/dog.jpg',cv2.IMREAD_GRAYSCALE)

src1=cv2.resize(src1,(500,500))
src2=cv2.resize(src2,(500,500))

dst=cv2.subtract(src1,src2)

plt.subplot(131), plt.axis('off'), plt.imshow(src1,cmap='gray'), plt.title('cat')
plt.subplot(132), plt.axis('off'), plt.imshow(src2,cmap='gray'), plt.title('dog')
plt.subplot(133), plt.axis('off'), plt.imshow(dst,cmap='gray'), plt.title('subtarct')
```

### 차이 연산
- 두 영상의 같은 위치에 존재하는 픽셀 값에 대해 뺄셈 연산 수행 후, 절댓값을 결과 영상의 픽셀값으로 설정
- $\text{dst}(x,y) = \text{saturate}(|\text{src1}(x,y) - \text{src2}(x,y)|)$
- 입력 영상 순서 영향 x
- cv2.absdiff(src1, src2, dst = None)
    - src1 : 첫번째 영상 or 스칼라
    - src2 : 두번째 영상 or 스칼라
    - dst : 차이 연산 결과

```python
import cv2
from matplotlib import pyplot as plt

src1 = cv2.imread('./img/abs1.jpg', cv2.IMREAD_GRAYSCALE)
src2 = cv2.imread('./img/abs2.jpg', cv2.IMREAD_GRAYSCALE)
dst = cv2.absdiff(src1, src2)

plt.subplot(131), plt.axis('off'), plt.imshow(src1, 'gray'), plt.title('abs1')
plt.subplot(132), plt.axis('off'), plt.imshow(src2, 'gray'), plt.title('abs2')
plt.subplot(133), plt.axis('off'), plt.imshow(dst, 'gray'), plt.title('result')
plt.show()
```

### 색상(채널) 분리
- RGB 색상 평면 분리
- cv2.split(src, mv=None)
    - src : 다채널 영상 -> (B,G,R)

```python
import cv2
from matplotlib import pyplot as plt

image = cv2.imread('./img/candies.png')
b, g, r = cv2.split(image)
merge_b = cv2.merge([b, b, b])
merge_g = cv2.merge([g, g, g])
merge_r= cv2.merge([r, r, r])

plt.subplot(221), plt.axis('off'), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.title('Original Image')
plt.subplot(222), plt.axis('off'), plt.imshow(merge_b, cmap='gray'), plt.title('Blue Channel')
plt.subplot(223), plt.axis('off'), plt.imshow(merge_g, cmap='gray'), plt.title('Green Channel')
plt.subplot(224), plt.axis('off'), plt.imshow(merge_r, cmap='gray'), plt.title('Red Channel')
plt.show()
```

### HSV 색공간
- 컬러 정의 방법 중 한 방식
    - Hue(색상) : 색 종류, 0-179
    - Saturartion(채도) : 색 선명 정도, 0-255

```python
image = cv2.imread('./img/candies.png')
hsv_image = cv2.cvtColor(image,cv2.COLOR_BGR2HSV)
h,s,v = cv2.split(hsv_image)

merge_h = cv2.merge([h,h,h])
merge_s = cv2.merge([s,s,s])
merge_v = cv2.merge([v,v,v])

plt.subplot(221), plt.axis('off'), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.title('Original Image')
plt.subplot(222), plt.axis('off'), plt.imshow(merge_h, cmap='gray'), plt.title('Hue Channel')
plt.subplot(223), plt.axis('off'), plt.imshow(merge_s, cmap='gray'), plt.title('Saturation Channel')
plt.subplot(224), plt.axis('off'), plt.imshow(merge_v, cmap='gray'), plt.title('Value Channel')
plt.show()
```

### 영상의 이진화(Binarization)
- 이진화(Binarization)은 두부분으로 나누는 것을 의미
- 디지털 영상의 이진화는 영상의 픽셀을 두 종류로 분류하는 처리
- 디지털 영상에서 객체 영역 (관심 영역)과 배경 영역(비관심 영역)으로 나누는 용도로 사용
- 이진화는 픽셀 값을 (검은색) 또는 255(흰색)로 설정



### 그레이스케일 영상의 이진화
- 그레이스케일 영상의 이진화는 픽셀 값이 주어진 경계값(임계값)을 기준으로 0(검은색)과 255(흰색)으로 분류하여 표현
- 아래 이미지(a)에 대한 히스토그램(b)을 분석해 보면 픽셀 값이 90, 210 두 지점에 모여 있음
- 경계값(임계값)을 150로 설정한다면 두 지점을 구분 가능하여 이진화를 진행(c)



- cv2.threshold(src, thresh, maxval, type, dst=None)
    - src : 입력 영상 -> 다채널, 8비트 또는 32비트 실수형
    - thresh : 사용자 지정 임계값
    - maxval : cv2.THRESH_BINARY 또는 cv2.THRESH_BINARY_INV 방법 사용 시 최대값 보통 255로 지정
    - type : cv2.THRESH_로 시작하는 플래그 -> 임계값 함수 동작 지정 또는 자동 임계값 결정 방법 지정
    - dst : 출력 영상 -> src와 동일 크기, 동일 타입, 같은 채널 수

```python
import cv2, sys
import numpy as np

src = cv2.imread('./img/neutrophils.png', cv2.IMREAD_GRAYSCALE)
def on_threshold(pos):
  _,dst= cv2.threshold (src, pos, 255, cv2. THRESH_BINARY)
  cv2.imshow('dst', dst)

cv2.imshow('src', src)
cv2.namedWindow('dst')
cv2.createTrackbar ('Threshold', 'dst', 0, 255, on_threshold)
cv2.setTrackbarPos ('Threshold', 'dst', 128)
cv2.waitKey()
cv2.destroyAllWindows()
```

# 레이블링
### 레이블링(labeling) 처리 과정
- 이진화된 디지털 영상에서 배경은 검은색 픽셀, 객체는 흰색 픽셀로 정의 
    - 디지털 영상에서 픽셀 값이 0이면 배경, 0이 아니면 객체 픽셀로 인식
- 객체는 한 개 이상의 연결된 픽셀로 이루어짐
- 객체를 구성하는 모든 픽셀에는 같은 레이블 번호를 부여
- 최종적으로 객체별로 레이블 번호는 다르게 지정


- 레이블링 처리후에는 객체별 영역에 객체별 레이블링 번호가 부여된 2차원 배열이 생성됨
- 이렇게 생성된 2차원 배열을 레이블 맵(label map)이라고 부름
- 레이블링 맵에 저장된 값은 정수로 저장
- 레이블링을 수행하는 알고리즘은 매우 다양하게 존재하지만 모두 같은 형태의 레이블맵을 생성함



- cv2.connectedComponentsWithStats (image, labels=None, stats = None, centroids = None, connectivity = None, Itype = None)
    - image : 8비트 1채널 영상
    - labels : 레이블 맵 행렬. 입력 영상과 같은 크기 -> numpy.ndarray
    - stats : 각 객체의 바운딩 박스, 픽셀 개수 정보를 담은 행렬
        - numpy.ndarray. shape=(N, 5), dtype=numpy.int32
    - centroids : 각 객체의 무게 중심 위치 정보를 담은 행렬
        - numpy.ndarray. shape=(N, 2), dtype=numpy.float64
    - connectivity : 4 또는 8 (기본값 = 8)
    - Itype : labels 타입 -> cv2.CV_32S 또는 cv2.CV_16S (기본값 = cv2.CV_32S) 

```python
import cv2, sys
src= cv2.imread('./img/keyboard.png', cv2.IMREAD_GRAYSCALE)
_, src_bin = cv2.threshold (src, 0, 255, cv2.THRESH_OTSU)
cnt, labels, stats, centroids = cv2.connectedComponentsWithStats (src_bin)
dst = cv2.cvtColor(src, cv2.COLOR_GRAY2BGR)

for i in range(1, cnt):
  (x, y, w, h, area) = stats[i]
  if area < 20:
    continue
  cv2.rectangle(dst, (x, y, w, h), (0, 255, 255))

cv2.imshow('src', src)
cv2.imshow('src_bin', src_bin)
cv2.imshow('dst', dst)
cv2.waitKey()
cv2.destroyAllWindows()
```

### 외곽선 검출
- 외곽선(contour) 검출이란 객체의 외곽선 좌표를 모두 추출하는 작업
    - 바깥선 외곽선뿐 아니라 옵션을 적용하여 안쪽 외곽선도 검출 가능
    - 외곽선 간의 상관관계도 알 수 있으며, 외곽선의 계층 구조도 표현 가능
    - 여러 외곽선을 표현할 때는 객체 하나의 외곽선을 원소로 갖는 리스트로 표현
    - 전체 리스트의 갯수는 전체 외곽선 갯수(N)
    
- cv2.findContours(image, mode, method, contours = None, hierarchy = None, offset = None)
    - image : 입력 영상 (non-zero 픽셀을 객체로 간주함)
    - mode : 외곽선 검출 모드 -> cv2.RETR_로 시작하는 상수
    - method : 외곽선 근사화 방법 -> CV2.CHAIN_APPROX_로 시작하는 상수
    - contours : 검출된 외곽선 좌표 -> numpy.ndarray로 구성된 리스트
    - hierarchy : 외곽선 계층 정보 -> numpy.ndarray. shape=(1, N, 4)
    - offset : 좌표 값 이동 옵셋 (기본값 = (0, 0))

- cv2.drawContours(image, contours, contourldx, color, thickness = None, lineType = None, hierarchy = None, maxLevel = None)
    - image : 입력 영상 (non-zero 픽셀을 객체로 간주함)
    - contours : (cv2.findContours() 함수로 구한) 외곽선 좌표 정보
    - contourldx : 외곽선 인덱스 -> 음수(-1)를 지정하면 모든 외곽선을 그림
    - color : 외곽선 색상
    - thickness : 외곽선 두께 -> thinkness < 0이면 내부를 채움
    - lineType : LINE_4, LINE_8, LINE_AA 중 하나 지정
    - hierarchy : 외곽선 계층 정보
    - maxLevel : 그리기를 수행할 최대 외곽선 레벨 -> maxLevel=0 이면 contourldx로 지정된 외곽선만 그림

```python
import cv2, random
src = cv2.imread('./img/contours.bmp', cv2.IMREAD_GRAYSCALE)
contours, hier=cv2.findContours (src, cv2.RETR_CCOMP, cv2.CHAIN_APPROX_NONE)
dst = cv2.cvtColor(src, cv2.COLOR_GRAY2BGR)
idx = 0

while idx >=0:
    c = (random.randint(0,255), random.randint(0,255), random.randint(0,255))
    cv2.drawContours (dst, contours, idx, c, 2, cv2.LINE_8, hier)
    idx = hier [0, idx, 0]

cv2.imshow('src', src)
cv2.imshow('dst', dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
